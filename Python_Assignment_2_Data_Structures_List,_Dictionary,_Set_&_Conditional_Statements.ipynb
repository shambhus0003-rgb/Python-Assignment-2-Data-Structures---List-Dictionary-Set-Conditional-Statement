```python
# 1. List Creation
age_list = [24, 25, 26, 27, 28]
name_list = ["Arjun", "Priya", "Karthik", "Divya", "Sneha"]

# 2. List Operations / Modifications
# a. Append "Yazhini" to name_list
name_list.append("Yazhini")
print("After append:", name_list)

# b. Insert 30 at index 2 in age_list
age_list.insert(2, 30)
print("After insert:", age_list)

# c. Remove "Yazhini" from name_list
name_list.remove("Yazhini")
print("After remove:", name_list)

# d. Pop the last element from age_list
popped_age = age_list.pop()
print("Popped element:", popped_age)
print("After pop:", age_list)

# e. Extend age_list with [29, 30, 26]
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# f. Sort age_list in descending order
age_list.sort(reverse=True)
print("Descending sort:", age_list)

# g. Find Max, Min, and Sum of age_list
print("Max age:", max(age_list))
print("Min age:", min(age_list))
print("Sum of ages:", sum(age_list))

# 3. Accessing List Elements
# a. First element
print("First element:", name_list[0])

# b. Last element
print("Last element:", name_list[-1])

# c. Elements from index 2 to index 4 (inclusive of index 4)
print("Index 2 to 4:", name_list[2:5])

# d. Elements in reverse order
print("Reverse order:", name_list[::-1])
```

### Explanation: List Operations & Access
* **Creation:** Lists are ordered, mutable collections created using square brackets `[]`.
* **Modifications:**
  * `.append()` adds a single item to the end of the list.
  * `.insert(index, value)` places a value at a specific position and shifts existing elements to the right.
  * `.remove(value)` deletes the first matching occurrence of the specified value.
  * `.pop()` removes and returns the last item in the list when no index is specified.
  * `.extend()` unpacks another iterable and appends all its elements to the end of the list.
  * `.sort(reverse=True)` arranges elements in descending order in-place.
  * Built-in functions `max()`, `min()`, and `sum()` compute the largest value, smallest value, and total sum of numeric elements.
* **Indexing & Slicing:**
  * Python uses zero-based indexing (`[0]` for the first item) and negative indexing (`[-1]` for the last item).
  * Slicing `name_list[2:5]` extracts elements from index 2 up to index 4 (the stop index `5` is excluded).
  * `name_list[::-1]` uses a step of `-1` to return a reversed copy of the list.

```python
# a. Create dictionary mapping 5 students to marks
student_marks = {
    "Rohan": 88,
    "Meera": 92,
    "Vikram": 76,
    "Ananya": 85,
    "Siddharth": 90
}

# b. Access and print the mark of a specific student
print("Meera's mark:", student_marks["Meera"])

# c. Add a new student "Janani" with a mark of 80
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# d. Update the mark of an older student to 82
student_marks["Vikram"] = 82
print("After updating Vikram's mark:", student_marks)

# e. Print all keys, values, and key-value pairs
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Items:", student_marks.items())
```

### Explanation: Dictionary Operations
* **Creation & Access:** Dictionaries store data as `key: value` pairs inside curly braces `{}`. Values are retrieved directly using their unique keys (`student_marks["Meera"]`).
* **Adding & Updating:** Assigning a value to a key that does not exist (`student_marks["Janani"] = 80`) inserts a new key-value pair. Assigning a value to an existing key (`student_marks["Vikram"] = 82`) overwrites its old value.
* **Dictionary View Methods:**
  * `.keys()` returns a view object containing all student names.
  * `.values()` returns a view object containing all marks.
  * `.items()` returns a view object of `(key, value)` tuples representing every entry in the dictionary.

```python
# a. Create a set with duplicate vowels
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print("my_set output:", my_set)



# c. Create set1 and set2
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# d. Compute and print union and intersection
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

### Explanation: Set Operations & Properties
* **Duplicate Removal (`my_set`):** Sets only store unique elements. When initialized with `['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i']`, Python automatically removes the duplicate `'a'` and `'i'` entries, leaving only the 5 unique vowels: `{'a', 'e', 'i', 'o', 'u'}` (displayed in arbitrary order).
* **Why `my_set[4] = 's'` Throws an Error:** Running this line raises a **`TypeError: 'set' `**. Sets are **unordered** collections implemented using hash tables, meaning elements have no fixed index positions. Because items cannot be accessed or assigned by index, attempting `my_set[4]` fails.
* **Union & Intersection:**
  * **Union (`.union()` or `|`):** Combines all unique elements present in either `set1` or `set2` (`{1, 2, 3, 5, 7, 8, 9, 10}`).
  * **Intersection (`.intersection()` or `&`):** Extracts only the elements common to both sets (`{3, 5}`).

```python
# 1. Prompt user for input (0 to 10 inclusive)
score = float(input("Enter your score (0 to 10): "))

# 2, 3 & 4. Evaluate performance category and print custom message
if score < 0 or score > 10:
    print("Invalid input! Please enter a score between 0 and 10.")
elif score > 7:
    print("Above Average: Excellent work! Keep up the great momentum.")
elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")
else:
    print("Below Average: Need to improve your performance, consistent practice will lead to better results.")
```

### Explanation: Performance Category Program
* **Input Handling:** `input()` captures the user's entry from the console, and `float()` converts it into a number so both whole numbers (e.g., `7`) and decimals (e.g., `7.5`) can be compared accurately.
* **Input Validation:** The initial `if score < 0 or score > 10` check ensures the score stays within the valid `0` to `10` boundary before categorizing.
* **Conditional Branching (`elif` / `else`):**
  * `elif score > 7:` triggers for any valid score strictly greater than `7` (up to `10`) and prints the **Above Average** message.
  * `elif score >= 4:` runs only if the previous condition was false (meaning the score is `7` or less). Checking `>= 4` captures all scores between `4` and `7` (inclusive) for the **Average** category.
  * `else:` catches all remaining valid numbers (scores strictly less than `4` down to `0`) and displays the **Below Average** feedback.
